```python
import os
import shutil
from  pathlib import Path
import torch
from safetensors.torch import save_model

def model_artifacts_ready():
    hugging_dir = Path("peak_artifacts")
    tokenizer_dir = Path("exp_1/tokenizer")
    weights_dir = Path("weights")
    if os.path.isdir(hugging_dir):
        shutil.rmtree(hugging_dir)
        os.makedirs(hugging_dir)
    else:
        os.makedirs(hugging_dir)

    if os.path.isdir(tokenizer_dir):
        for file in os.listdir(tokenizer_dir):
            shutil.copy2(tokenizer_dir / file, hugging_dir / file)
    else:
        raise FileNotFoundError("Tokenizer directory not found")

    if os.path.isdir(weights_dir):
        for file in os.listdir(weights_dir):
            shutil.copy2(weights_dir / file, hugging_dir / file)
    else:
        raise FileNotFoundError("Weight directory not found")

# model_artifacts_ready()
```
